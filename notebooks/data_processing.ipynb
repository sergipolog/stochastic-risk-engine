## DATA EXPLORATION

This section shows the inital data exploration and cleaning.

```python
import pandas as pd
```

```python
data = pd.read_csv('../data/raw/historical_stock.csv', index_col=0, header=[0,1])

data.columns = [f"{col[0]}_{col[1]}" for col in data.columns]

df = pd.DataFrame(data)

df.head(10)
```

```python
df.info() 
df.isnull().sum()
```

The *historical_stock.csv* file has 775 entries, from **2022-01-01** to **2025-01-01**. It is easyly observable that at least 7 days are null in every column, those days need to be assessed, as well as some other faulty registries clearly shown in the information above. 

```python
null_dates = df.isnull()

null_dates = null_dates[null_dates.any(axis=1)]
null_dates = null_dates.index.tolist()
print(null_dates)
print(f"Number of null dates: {len(null_dates)}")
print(f"Percentage of null dates: {len(null_dates)/len(df)*100:.2f}%")
```

Being the percentage of null dates as low as $4\%$, these registries can be deleted without compromising the integrity of this study. 

```python
df.dropna(inplace=True)

df.to_csv('../data/processed/historical_stock_cleaned.csv')
```
